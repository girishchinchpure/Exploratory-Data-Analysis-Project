```python
print(df.shape)
print()
df.info()
print()
df.describe(include="all")
```

```python
# Handling Missing Values
# Year and publisher are two columns which have null values
# Year - 271 missing values
# Publisher - 58 missing values
# Both the columns have very less percentage of missing values,so we will use dropna() function as it won't affect overall data

df = df.dropna(subset=['Publisher']) #Deletes all rows of the column "Year" having missing values
df = df.dropna(subset=['Year']) #Deletes all rows of the column "Year" having missing values

df.info() # Verification of handled missing values

print(df.columns) # No Leading/Trailing spaces
```

```python
# Handling Duplicate values
print(df.duplicated().sum())
print()
# if duplicates existed, then "print(df.drop_duplicates(inplace=True))" would be used, it Deletes duplicate rows

df['Year'] = df['Year'].astype(int) # Changing data type of "Year" from float to integer
df.info()
print()

df['Decade'] = (df['Year']//10)*10 # Adding Columns which can be used in EDA
df.head(10)
```

Visualisation

```python
platform_values = df['Platform'].value_counts().head(10) # Taking the top 10 platforms because there are too many platforms
plt.figure(figsize=(14,7))

x = sns.barplot(
    x = platform_values.index,
    y = platform_values.values
)
for i in x.containers:
    x.bar_label(i)

plt.title("Number of Games on platforms")
plt.xlabel("Platforms")
plt.ylabel("Number of Games")
plt.show()

plt.savefig(
    "C:\\Users\\Supriya Chinchpure\\Desktop\\Girish\\Exploratory Data Analysis (EDA) Project\\Visuals\\01_top_platforms.png",
    bbox_inches = 'tight',
    dpi = 300
)
```

```python
genres = df['Genre'].value_counts()
plt.figure(figsize=(14,7))
x = sns.barplot(
    x = genres.index,
    y = genres.values
)
plt.title("Top Genres")
plt.xlabel("Genres")
plt.ylabel("Number of Games")

for i in x.containers:
    x.bar_label(i)

plt.show()

plt.savefig(
    "C:\\Users\\Supriya Chinchpure\\Desktop\\Girish\\Exploratory Data Analysis (EDA) Project\\Visuals\\02_top_genres.png",
    bbox_inches = 'tight',
    dpi = 300
)
```

```python
released = df['Year'].value_counts().sort_index()
plt.figure(figsize=(14,7))

x = sns.lineplot(
    x = released.index,
    y = released.values,
)

plt.title("Games Released per Year")
plt.xlabel("Years")
plt.ylabel("Number of Games")

plt.savefig(
    "C:\\Users\\Supriya Chinchpure\\Desktop\\Girish\\Exploratory Data Analysis (EDA) Project\\Visuals\\03_games_released.png",
    bbox_inches = 'tight',
    dpi = 300
)

plt.show()
```
